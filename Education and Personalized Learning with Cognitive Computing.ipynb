```python
# ✅ Session 8: Personalized Learning with Cognitive Computing

# 📌 Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 📌 Step 2: Simulated Student Performance Dataset
data = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Math': [78, 45, 62, 90, 55, 72, 33, 84],
    'Science': [82, 48, 60, 88, 50, 69, 39, 86],
    'English': [75, 52, 58, 92, 45, 74, 35, 80]
}
df = pd.DataFrame(data)
df.set_index('Student', inplace=True)

# 📌 Step 3: Normalize the Data
scaler = StandardScaler()
scaled = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled, index=df.index, columns=df.columns)

# 📌 Step 4: Apply Clustering (KMeans)
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(scaled_df)
df['Learning_Profile'] = clusters

# 📌 Step 5: Display Clustered Data
print("Clustered Student Data:")
print(df)

# 📊 Step 6: Visualize Clusters
plt.figure(figsize=(8,6))
sns.scatterplot(x='Math', y='Science', hue='Learning_Profile', data=df, palette='Set2', s=100)
plt.title("Student Clusters Based on Learning Performance")
plt.xlabel("Math Score")
plt.ylabel("Science Score")
plt.grid(True)
plt.show()

# 📌 Step 7: Adaptive Learning Recommendations
def recommend_learning_path(profile_id):
    if profile_id == 0:
        return "🟡 Focus on strengthening foundational concepts."
    elif profile_id == 1:
        return "🟢 Ready for advanced materials and real-world projects."
    elif profile_id == 2:
        return "🔴 Needs additional support and guided tutorials."
    else:
        return "Invalid profile."

# 📌 Step 8: Recommend for Each Student
print("\n📚 Personalized Learning Recommendations:")
for student in df.index:
    profile = df.loc[student, 'Learning_Profile']
    recommendation = recommend_learning_path(profile)
    print(f"{student}: {recommendation}")
```
